```python
#12.Data_Aggregation.ipynb
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Import dataset
df = pd.read_csv("Retail_Sales_Data.csv")

# 2. Explore dataset
print(df.head())
print(df.info())
print(df.describe())

# 3. Identify relevant variables
print("Columns:", df.columns.tolist())

# 4. Aggregate sales by region
region_sales = df.groupby("Region")["Sales_Amount"].sum()
print("\nTotal Sales by Region:")
print(region_sales)

# 5. Visualize regional sales
region_sales.plot(kind="bar", title="Sales by Region")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()

# 6. Identify top-performing region
print("\nTop Performing Region:")
print(region_sales.idxmax(), region_sales.max())

# 7. Aggregate by region and product category
category_sales = df.pivot_table(
    index="Region",
    columns="Product_Category",
    values="Sales_Amount",
    aggfunc="sum",
    fill_value=0
)
print("\nSales by Region and Category:")
print(category_sales)

# 8. Create stacked bar plot
category_sales.plot(kind="bar", stacked=True)
plt.title("Regional Sales by Product Category")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```
